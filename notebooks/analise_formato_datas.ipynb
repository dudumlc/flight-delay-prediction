```python
df = pd.read_sql_query("SELECT * FROM brz_clima", conn)
```

```python
df['anoRef'] = df['anoRef'].astype(int)
df['data'] = pd.to_datetime(df['data'], errors='coerce')
df['mesRef'] = df['data'].dt.month
df = df.replace('nan', pd.NA)

df = df[df['anoRef'] >= 2021]
```
