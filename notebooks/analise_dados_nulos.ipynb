```python
import pandas as pd
import numpy as np

def avaliar_lacunas(df):
    # Garante ordenação
    df = df.sort_values(by=['data', 'hora'])
    
    # Seleciona apenas as colunas de dados
    num_cols = df.iloc[:, 2:-2].columns
    
    print(f"{'Coluna':<25} | {'Maior Sequência Nula':<20} | {'Total de Nulos'}")
    print("-" * 65)
    
    for col in num_cols:
        # Cria uma série booleana onde True é nulo
        isna = df[col].isna()
        
        # Lógica de agrupamento: incrementa o grupo toda vez que o status (nulo/não nulo) muda
        grupos = (isna != isna.shift()).cumsum()
        
        # Filtra apenas onde é nulo e conta o tamanho de cada sequência
        sequencias_nulas = isna[isna].groupby(grupos).sum()
        
        if not sequencias_nulas.empty:
            maior_buraco = sequencias_nulas.max()
            total_nulos = isna.sum()
        else:
            maior_buraco = 0
            total_nulos = 0
            
        print(f"{col:<25} | {int(maior_buraco):>2} registros {'':<8} | {total_nulos:>5}")

# Exemplo de uso:
# avaliar_lacunas(seu_dataframe)
```

```python
avaliar_lacunas(df)
```
