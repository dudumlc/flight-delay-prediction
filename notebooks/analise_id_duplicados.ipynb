```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Pichau\Downloads\idVoo,empresaAerea,numeroVoo,codeAutoriz.txt")
```

```python
df[df['numeroVoo'] == 2419].sort_values('numeroVoo', ascending=True)
```

```python
df['linhaDuplicada'] = df.duplicated(keep=False)
```

```python
# Seleciona as duas linhas para comparação
linha_0 = df.iloc[1]
linha_2 = df.iloc[2]

# Itera sobre as colunas e mostra onde elas divergem
for col in df.columns:
    val0 = linha_0[col]
    val2 = linha_2[col]
    if val0 != val2:
        print(f"Diferença encontrada na coluna: {col}")
        print(f"Linha 0: '{val0}' (Tipo: {type(val0)})")
        print(f"Linha 2: '{val2}' (Tipo: {type(val2)})")
        print("-" * 30)
    else:
        print(f'Coluna {col}: tudo certo!')
```

```python
# Compara as linhas usando o método equals do Pandas (mais rigoroso)
print(f"Linha 0 é idêntica à 2? {df.iloc[0:1].equals(df.iloc[2:3])}")

# Mostra se há NaNs escondidos que podem estar quebrando a lógica
print("\nContagem de nulos nas linhas:")
print(df.iloc[[0, 2]].isna().sum(axis=1))

# Tente converter tudo para string temporariamente e testar
temp_duplicados = df.astype(str).duplicated(keep=False)
print(f"\nSe eu converter tudo para STRING, a linha 2 vira duplicada? {temp_duplicados.iloc[2]}")
```
